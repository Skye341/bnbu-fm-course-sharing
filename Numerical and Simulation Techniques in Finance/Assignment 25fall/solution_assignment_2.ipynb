Consider the following first-order partial differential equation

### **Question 1**  &nbsp;&nbsp; (5 points)

Verify that the analytical solution of the PDE is given by 

$${\large\left.\begin{array}{lll}
u(t,x)=f(x-at) & , & t{\geq}0,\ x\in\mathbb{R}\end{array}\right.}$$  

### **Answer 1**  
Define

$${\large\left.\begin{array}{lll}
u(t,x)=f(x-at) & , & t{\geq}0,\ x\in\mathbb{R}\end{array}\right.}$$ 

The partial derivatives and initial values of $u$ are given by

$${\large\begin{array}{ccccc}
\frac{\partial u}{\partial t}(t,x)=-af'(x-at) &  & \frac{\partial u}{\partial x}(t,x)=f'(x-at) &  & u(0,x)=f(x)\end{array}}$$

Therefore it is true that $\frac{\partial u}{\partial t}+a\frac{\partial u}{\partial x}=0$, which means that $u(t,x)=f(x-at)$ solves the PDE and has the proper initial condition.  

---
We are going to approximate the solution of the PDE by finite-difference schemes. Introduce the uniform grids
$${\large\begin{array}{lll}
t_{i}=i\times\Delta_{t}\ , & i=0,1,\ldots,N & \mathrm{where}\ \Delta_{t}=\frac{t_{N}}{N}\\
x_{j}=x_{0}+j\times\Delta_{x}\ , & j=0,1,\ldots,M & \mathrm{where}\ \Delta_{x}=\frac{x_{M}-x_{0}}{M}
\end{array}}$$

Denote $u_{i,j}$ as the approximation of the solution $u(t_i,x_j)$ at time $t_i$ and space $x_j$. Similarly, define $f_j=f(x_j)$ for all $j=0,1,\ldots,M$.

### **Answer 4**  
Below is an implementation of the FTBS finite-difference scheme in Python:


```python
import numpy as np
from scipy.sparse import diags

def FTBS_scheme(x0,xM,tN,f,a,N,M):
  dt = tN/N # constant step size of the time grid
  dx = (xM-x0)/M # constant step size of the space grid
  tgrid = np.linspace(0.0,tN,N+1) # time grid
  xgrid = np.linspace(x0,xM,M+1) # space grid
  lmd = a*dt/dx # lambda constant in the FTBS scheme
  m = np.repeat(1-lmd,M+1) # diagonal of Ar matrix
  d = np.repeat(lmd,M) # lower diagonal of Ar matrix
  Ar = diags([d, m], [-1, 0]).toarray() # Ar matrix
  Br = np.zeros(M+1) # Br vector
  Br[0] = lmd # Br is zero everywhere except the first element
  U = np.zeros([N+1,M+1]) # the matrix U will contain the FTBS scheme results
  U[0,:] = f(xgrid) # initial condition
  Ubound = f(x0-dx-a*tgrid) # boundary condition provided by analytical solution
  for i in np.arange(N): 
      U[i+1,:] = Ar@U[i,:] + Br*Ubound[i] # FTBS scheme 
  return U[-1,:] # return the terminal values
```

```python
# parameters
x0, xM = -4.0, 4.0
tN, a = 2.0, 1.0

scheme = 'FTBS' # name of finite-difference scheme
title = scheme+' scheme'
fig, ax = plt.subplots()
oneline(0,1000,"exact",ax)
oneline(500,250,scheme,ax)
oneline(150,75,scheme,ax)
oneline(50,25,scheme,ax)
ax.set_xlabel("$x$", loc='right')
plt.title(title, loc='left')
plt.legend(loc='upper left')
plt.show()
```

#### *More information about this PDE*:  
With the first-order hyperbolic PDE considered in this assignment, the FTBS explicit scheme is only conditionally stable, and the stability condition is 

$${\large\left|\lambda\right|=\left|a\frac{\Delta_{t}}{\Delta_{x}}\right|\leq1}$$
This inequality is called the **Courant-Friedrichs-Lewy (CFL)** condition, and is one of the most famous inequalities in numerical analysis. In this example, the CFL condition means that $N$ needs to be greater than $M$ for the FTBS explicit scheme to converge to the analytical solution.

### **Question 8**  &nbsp;&nbsp; (7 points)

Once again, use the analytical solution to define the boundary condition $u_{i+1,-1}$. Implement the BTBS finite-difference scheme in Python. Use the same numerical parameters as in question 4. Plot the solution you obtain for $\{u_{N,j}\}_{j=0,\ldots,M}$. It should look like the following:



### **Answer 8**  

Below is an implementation of the BTBS finite-difference scheme in Python:


```python
import numpy as np
from scipy.sparse import diags
from scipy.linalg import solve

def BTBS_scheme(x0,xM,tN,f,a,N,M):
  dt = tN/N # constant step size of the time grid
  dx = (xM-x0)/M # constant step size of the space grid
  tgrid = np.linspace(0.0,tN,N+1) # time grid
  xgrid = np.linspace(x0,xM,M+1) # space grid
  lmd = a*dt/dx # lambda constant in the BTBS scheme
  m = np.repeat(1+lmd,M+1) # diagonal of Al matrix
  d = np.repeat(-lmd,M) # lower diagonal of Al matrix
  Al = diags([d, m], [-1, 0]).toarray() #  Al matrix
  Bl = np.zeros(M+1) # Bl vector
  Bl[0] = lmd # Bl is zero everywhere except the first element
  U = np.zeros([N+1,M+1]) # the matrix U will contain the BTBS scheme results
  U[0,:] = f(xgrid) # initial condition
  Ubound = f(x0-dx-a*tgrid) # boundary condition provided by analytical solution
  for i in np.arange(N):
      U[i+1,:] = solve(Al, U[i,:]+Bl*Ubound[i+1]) # BTBS scheme 
  return U[-1,:] # return the terminal values
```

What happens to your numerical solution if you choose $M$ much larger than $N$, for example $N=50$ and $M=500$?  
Set $M=2{\times}N$, and find the lowest $N>0$ such that the maximum absolute error between the analytical solution at time $t_N$ and the BTBS scheme solution at time $t_N$ on the grid $\{x_{j}\}_{j=0,1,\ldots,M}$ is lower than $0.02$.  

### **Answer 9**  

With the BTBS scheme, there is no numerical stability problem when $M$ is much larger than $N$. Indeed, the BTBS scheme is an **implicit scheme**, which is always **unconditionally stable**. The results for $N=50$ and $M=500$ are shown below



Below is the code to find the lowest $N>0$ such that the maximum absolute error between the analytical solution at time $t_N$ and the BTBS scheme solution at time $t_N$ on the grid $\{x_{j}\}_{j=0,1,\ldots,M}$ is lower than $0.02$


```python
N = 2; M = 2*N
xgrid = np.linspace(x0,xM,M+1)
U_exact = cauchy.pdf(xgrid-a*tN)
U_btbs = BTBS_scheme(x0,xM,tN,cauchy.pdf,a,N,M)
error = np.max(np.abs(U_exact-U_btbs))
while (error>0.02):
    N += 1; M = 2*N
    xgrid = np.linspace(x0,xM,M+1)
    U_exact = cauchy.pdf(xgrid-a*tN)
    U_btbs = BTBS_scheme(x0,xM,tN,cauchy.pdf,a,N,M)
    error = np.max(np.abs(U_exact-U_btbs))

print("N =",N,", error =",error)
```
